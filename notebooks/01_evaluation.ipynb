# AR-QF evaluation
Run the experiment, inspect results, compute concentration and generate figures.

```python
from arqf.config import ExperimentConfig
from arqf.experiment import run
cfg=ExperimentConfig()
run(cfg,'outputs')
```

```python
import pandas as pd
df=pd.read_csv('outputs/results.csv')
df.groupby(['scenario','mechanism'])[['gini','contribution_gini','entropy','welfare']].mean()
```

```python
from arqf.analysis import make_figures
make_figures('outputs/results.csv','outputs/figures')
```
